# MEHLISSA Next: first scenario from Python

This notebook uses the UX-5 process API. The C++ executable and its JSON Schemas remain authoritative: Python launches the normal commands and reads their versioned results. The example is research software, not a clinical assay or medical device.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()  # start Jupyter in the repository root
sys.path.insert(0, str(ROOT / 'python'))
from mehlissa import MehlissaClient, load_result

candidates = [
    ROOT / 'build/windows-msvc/apps/Debug/mehlissa.exe',
    ROOT / 'build/windows-msvc/apps/Release/mehlissa.exe',
    ROOT / 'build/linux-gcc/apps/mehlissa',
]
executable = next(path for path in candidates if path.is_file())
client = MehlissaClient(executable, ROOT)
print(client.list_models(layer='organ'))
```

## Validate and run

The returned object contains exact paths printed by the C++ application, so no directory-name guessing is required.

```python
scenario_file = ROOT / 'examples/scenarios/fp9-lung-level-a-v1.json'
print(client.validate_scenario(scenario_file))
execution = client.run_scenario(scenario_file, ROOT / 'results/notebook-first-run')
execution.result
```

```python
result = load_result(execution.result)
result.summary
```

A positive software outcome does not establish sensitivity or specificity in patients. Always inspect `clinical_validation_claim` and the full `validity.limitations` array before interpreting or sharing a result.

```python
report_directory = ROOT / 'reports' / f'notebook-{execution.directory.name}'
report = client.report_result(execution.result, report_directory)
report.html
```
