# MEHLISSA Next: derived-experiment campaign

This notebook executes the checked-in UX-4 campaign and uses UX-5 helpers to group runs and calculate paired differences. These are descriptive software results, not a clinical sensitivity analysis.

```python
from datetime import datetime, timezone
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
sys.path.insert(0, str(ROOT / 'python'))
from mehlissa import MehlissaClient, load_campaign_result

candidates = [
    ROOT / 'build/windows-msvc/apps/Debug/mehlissa.exe',
    ROOT / 'build/windows-msvc/apps/Release/mehlissa.exe',
    ROOT / 'build/linux-gcc/apps/mehlissa',
]
executable = next(path for path in candidates if path.is_file())
client = MehlissaClient(executable, ROOT)
```

```python
campaign_file = ROOT / 'examples/campaigns/fp9-collector-count-v1.json'
print(client.validate_campaign(campaign_file))
stamp = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ')
execution = client.run_campaign(campaign_file, ROOT / 'results' / f'notebook-campaign-{stamp}')
execution.result
```

```python
campaign = load_campaign_result(execution.result)
{name: len(runs) for name, runs in campaign.groups().items()}
```

```python
campaign.paired_differences('sensitivity')
```

A zero paired difference here can reflect an insensitive or discrete demonstrator output; it is not evidence that collector count is biologically irrelevant. Inspect the campaign limitations and select better-qualified responses before scientific inference.
